```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import numpy as np
from tqdm.auto import tqdm
import torch

import autoencoder
#import run_experiment
import utils
import config
import training


def test_full_dictionary_tokenization(verbose=False):
    # load gpt2 tokenizer
    tokenizer = AutoTokenizer.from_pretrained("distilgpt2")
    # create a dictionary to store the results
    token_dict = {}
    for i in tqdm(range(1000)):  # assuming the range of tokens is from 0 to 1000
        # decode and encode the token
        sentence = tokenizer.decode([i])
        tokens = tokenizer.encode(sentence)
        # store the result in the dictionary
        token_dict[i] = tokens
        # check if the tokenization is the identity
        if i != tokens[0]:
            if verbose: print(f"Token {i} was changed to {tokens[0]} after decoding and encoding.")
    return token_dict

if __name__ == "__main__":
    token_dict = test_full_dictionary_tokenization()
    print(token_dict)
```

```python
# encode both keys and values
tokenizer = AutoTokenizer.from_pretrained("distilgpt2")
encoded_dict = {}
for key, value in token_dict.items():
    encoded_key = tokenizer.encode(str(key))
    encoded_value = [tokenizer.encode(str(v)) for v in value]
    encoded_dict[encoded_key[0]] = encoded_value
print(encoded_dict)
```
